```python
import pandas as pd
import glob

# Combine all pitch logs into 1 big dataframe
files = glob.glob("pitch_logs/**/*.csv", recursive=True)
df = pd.concat([pd.read_csv(f) for f in files], ignore_index=True)

len(df)
```

```python
df.columns
```

```python
pd.set_option('display.max_columns', None)
df.head()
```

```python
# Normalize all metrics to be from pitcher's perspective
df["plateLocX"] = -df["plateLocX"]
df["relX"] = -df["relX"]
```

```python
df.head()
```

```python
# Create fastball velo column, which takes the average velo for a pitcher's FF, SI, and FC for each season - to be used for Breaking Ball and Offspeed models
fb_velo = (
    df[df["pitchType"].isin(["FA", "SI", "FC"])]
    .groupby(["pitcherId", "year"])["releaseVelocity"]
    .mean()
    .reset_index(name="fb_velo")
)

df1 = df.merge(fb_velo, on=["pitcherId", "year"], how="left")
```

```python
df1["pitchType"].unique()
```

```python
# Assign categorical column for batter/pitcher handedness
df1["phand"] = (df1["pitcherHand"] != "L").astype(int)
df1["bhand"] = (df1["batterHand"] != "L").astype(int)
```

```python
df1.loc[df1["pitchType"].isin(["FA", "SI"]), "pitchBucket"] = "FB"
df1.loc[df1["pitchType"].isin(["FC"]), "pitchBucket"] = "FC"
df1.loc[df1["pitchType"].isin(["CH", "FS"]), "pitchBucket"] = "OFF"
df1.loc[df1["pitchType"].isin(["SL", "ST", "CU"]), "pitchBucket"] = "BB"
```

```python
df1 = df1.dropna()
```

```python
# Save as parquet (will use SQL to extract rows of interest moving forward)
df1.to_parquet("full_data.parquet")
```
